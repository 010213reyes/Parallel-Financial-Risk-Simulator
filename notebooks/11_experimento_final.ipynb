```python
best_parallel = (
    performance[
        performance["mode"] == "parallel"
    ]
    .sort_values(
        "elapsed_seconds"
    )
    .iloc[0]
)

print(
    "Mejor configuración paralela observada:"
)

print(
    f"Escenarios: "
    f"{int(best_parallel['scenarios']):,}"
)

print(
    f"Workers: "
    f"{int(best_parallel['workers'])}"
)

print(
    f"Tiempo: "
    f"{best_parallel['elapsed_seconds']:.4f} s"
)

print(
    f"Speedup: "
    f"{best_parallel['speedup']:.4f}"
)

print(
    f"Eficiencia: "
    f"{best_parallel['efficiency']:.4f}"
)
```

Por ello, los valores obtenidos corresponden al entorno

# Análisis adaptativo

Una característica diferenciadora del proyecto es el estudio de
un mecanismo capaz de modificar la cantidad de cálculo realizado
según la estabilidad estadística de los resultados.

En lugar de asumir que una cantidad fija de escenarios siempre es
suficiente, el simulador puede evaluar un criterio de convergencia
después de cada lote.

Esto abre la posibilidad de combinar:

    criterio estadístico
            +
    asignación dinámica de recursos
            +
    procesamiento paralelo
